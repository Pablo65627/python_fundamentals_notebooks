# Sección 19.1: Introducción al diccionario

Un diccionario es un ejemplo de almacén de clave-valor, también conocido como mapeo en Python. Permite almacenar y recuperar elementos mediante una clave. Dado que los diccionarios se referencian por clave, las búsquedas son muy rápidas. Sin embargo, como se utilizan principalmente para referenciar elementos por clave, no están ordenados.

creando un dictado

Los diccionarios se pueden iniciar de muchas maneras:

sintaxis literal

```python
d = {}                      # diccionario vacío
d = {'clave': 'valor'}      # diccionario con valores iniciales
```

Python 3.x Version ≥ 3.5

```python
# También es posible desempaquetar uno o varios diccionarios con la sintaxis literal.

# Crea una copia superficial de otherdict.
d = {**otherdict}
# También actualiza la copia superficial con el contenido de yetanotherdict.
d = {**otherdict, **yetanotherdict}
```

comprensión de diccionario

```python
d = {k:v for k,v in [('key', 'value',)]}
```

Véase también: Comprensiones

Clase integrada: dict()

```python
d = dict()                              # Diccionario vacío
d = dict(clave='valor')                 # Argumentos con nombre explícito
d = dict([('clave', 'valor')])          # Se pasa una lista de pares clave/valor
# Se crea una copia superficial de otro diccionario (¡solo es posible si las claves son cadenas de texto!)
d = dict(**otherdict)
```

modificar un diccionario

Para agregar elementos a un diccionario, simplemente cree una nueva clave con un valor:

```python
d['newkey'] = 42
```

También es posible agregar listas y diccionarios como valor:

```python
d['new_list'] = [1, 2, 3]
d['new_dict'] = {'nested_dict': 1}
```

Para eliminar un elemento, elimine la clave del diccionario:

```python
del d['newkey']
```

# Sección 19.2: Cómo evitar excepciones KeyError

Un error común al usar diccionarios es intentar acceder a una clave inexistente. Esto suele provocar una excepción KeyError.

```python
mydict = {}
mydict['not there']

Traceback (most recent call last):
 File "<stdin>", line 1, in <module>
KeyError: 'not there'
```

Una forma de evitar errores de clave es usar el método dict.get, que permite especificar un valor predeterminado que se devolverá en caso de que falte una clave.

```python
value = mydict.get(key, default_value)
```

Esto devuelve `mydict[key]` si existe, pero devuelve `default_value` en caso contrario. Ten en cuenta que esto no añade `key` a `mydict`. Por lo tanto, si quieres conservar ese par clave-valor, debes usar `mydict.setdefault(key, default_value)`, que sí almacena el par clave-valor.

```python
mydict = {}
print(mydict)
# {}
print(mydict.get("foo", "bar"))
# bar
print(mydict)
# {}
print(mydict.setdefault("foo", "bar"))
# bar
print(mydict)
# {'foo': 'bar'}
```

Una forma alternativa de abordar el problema es capturar la excepción.

```python
try:
    value = mydict[key]
except KeyError:
    value = default_value
```

También podrías comprobar si la clave está en el diccionario.

```python
if key in mydict:
    value = mydict[key]
else:
    value = default_value
```

Sin embargo, tenga en cuenta que en entornos multihilo es posible que la clave se elimine del diccionario después de la comprobación, lo que crea una condición de carrera en la que aún puede producirse la excepción.

Otra opción es usar una subclase de dict, collections.defaultdict, que tiene un default_factory para crear nuevas entradas en el diccionario cuando se le proporciona una nueva clave.

# Sección 19.3: Iteración sobre un diccionario

Si utilizas un diccionario como iterador (por ejemplo, en una instrucción for), este recorre las claves del diccionario. Por ejemplo:

```python
d = {'a': 1, 'b': 2, 'c':3}
for key in d:
    print(key, d[key])
# c 3
# b 2
# a 1
```

Lo mismo ocurre cuando se utiliza en una comprensión.

```python
print([key for key in d])
# ['c', 'b', 'a']
```

Python 3.x Version ≥ 3.0

El método items() se puede utilizar para iterar sobre la clave y el valor simultáneamente:

```python
for key, value in d.items():
    print(key, value)
# c 3
# b 2
# a 1
```

Si bien el método values() se puede usar para iterar solo sobre los valores, como cabría esperar:

```python
for key, value in d.values():
    print(key, value)
    # 3
    # 2
    # 1
```

Python 2.x Version ≥ 2.2

Aquí, los métodos keys(), values() e items() devuelven listas, y existen tres métodos adicionales, iterkeys(), itervalues() e iteritems(), para devolver iteradores.

# Sección 19.4: Diccionario con valores predeterminados

Disponible en la biblioteca estándar como [defaultdict](https://docs.python.org/3/library/collections.html#collections.defaultdict)

```python
from collections import defaultdict

d = defaultdict(int)
d['key']                            # 0
d['key'] = 5
d['key']                            # 5

d = defaultdict(lambda: 'empty')
d['key']                            # 'empty'
d['key'] = 'full'
d['key']                            # 'full'
```

[*] Como alternativa, si necesita usar la clase `dict` integrada, usar `dict.setdefault()` le permitirá crear un valor predeterminado cada vez que acceda a una clave que no existía previamente:

```python
>>> d = {}
{}
>>> d.setdefault('Another_key', []).append("This worked!")
>>> d
{'Another_key': ['This worked!']}
```

Ten en cuenta que si tienes muchos valores que añadir, `dict.setdefault()` creará una nueva instancia del valor inicial (en este ejemplo, un `[]`) cada vez que se llame, lo que puede generar una carga de trabajo innecesaria.

[*] Python Cookbook, 3.ª edición, de David Beazley y Brian K. Jones (O’Reilly). Copyright 2013 David Beazley y Brian Jones, 978-1-449-34037-7.

# Sección 19.5: Combinación de diccionarios

Consideremos los siguientes diccionarios:

```python
>>> fish = {'name': "Nemo", 'hands': "fins", 'special': "gills"}
>>> dog = {'name': "Clifford", 'hands': "paws", 'color': "red"}
```

Python 3.5+

```python
>>> fishdog = {**fish, **dog}
>>> fishdog
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Como demuestra este ejemplo, las claves duplicadas se asignan a su último valor (por ejemplo, "Clifford" sobrescribe a "Nemo").

Python 3.3+

```python
>>> from collections import ChainMap
>>> dict(ChainMap(fish, dog))
{'hands': 'fins', 'color': 'red', 'special': 'gills', 'name': 'Nemo'}
```

Con esta técnica, el valor más antiguo tiene prioridad para una clave dada en lugar del último ("Clifford" se descarta en favor de "Nemo").

Python 2.x, 3.x

```python
>>> from itertools import chain
>>> dict(chain(fish.items(), dog.items()))
{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Esto utiliza el último valor, como con la técnica basada en ** para la fusión ("Clifford" anula "Nemo").

```python
>>> fish.update(dog)
>>> fish
{'color': 'red', 'hands': 'paws', 'name': 'Clifford', 'special': 'gills'}
```

dict.update utiliza el último diccionario para sobrescribir el anterior.

# Sección 19.6: Acceso a claves y valores

Al trabajar con diccionarios, a menudo es necesario acceder a todas las claves y valores del diccionario, ya sea mediante un bucle for, una comprensión de lista o simplemente como una lista simple.

Dado un diccionario como este:

```python
mydict = {
    'a': '1',
    'b': '2'
}
```

Puedes obtener una lista de claves usando el método keys():

```python
print(mydict.keys())
# Python2: ['a', 'b']
# Python3: dict_keys(['b', 'a'])
```

Si en cambio desea una lista de valores, utilice el método values():

```python
print(mydict.values())
# Python2: ['1', '2']
# Python3: dict_values(['2', '1'])
```

Si desea trabajar tanto con la clave como con su valor correspondiente, puede utilizar el método items():

```python
print(mydict.items())
# Python2: [('a', '1'), ('b', '2')]
# Python3: dict_items([('b', '2'), ('a', '1')])
```

NOTA: Dado que un diccionario no está ordenado, las funciones keys(), values() e items() no tienen un orden de clasificación. Si le importa el orden en que devuelven estos métodos, utilice sort(), sorted() o un OrderedDict.

Diferencias entre Python 2 y 3: En Python 3, estos métodos devuelven objetos iterables especiales, no listas, y son el equivalente a los métodos `iterkeys()`, `itervalues()` e `iteritems()` de Python 2. Estos objetos se pueden usar como listas en la mayoría de los casos, aunque existen algunas diferencias. Consulte la [PEP 3106](https://www.python.org/dev/peps/pep-3106/) para obtener más detalles.

# Sección 19.7: Acceso a los valores de un diccionario

```python
dictionary = {"Hello": 1234, "World": 5678}
print(dictionary["Hello"])
```

El código anterior imprimirá 1234.

La cadena "Hello" en este ejemplo se llama key. Se utiliza para buscar un valor en el `dict` colocando la clave entre corchetes.

El número 1234 aparece después de los dos puntos correspondientes en la definición del `dict`. Este es el valor al que se asigna "Hello" en este `dict`.

Buscar un valor de este tipo con una clave inexistente generará una excepción KeyError, deteniendo la ejecución si no se controla. Si queremos acceder a un valor sin riesgo de una KeyError, podemos usar el método dictionary.get. Por defecto, si la clave no existe, el método devolverá None. Podemos pasarle un segundo valor para que devuelva None en caso de que la búsqueda falle.

```python
w = dictionary.get("whatever")
x = dictionary.get("whatever", "nuh-uh")
```

En este ejemplo, w obtendrá el valor None y x obtendrá el valor "nuh-uh".

# Sección 19.8: Creación de un diccionario

Reglas para crear un diccionario:  
  ·Cada clave debe ser única (de lo contrario, se sobrescribirá).  
  ·Cada clave debe ser hasheable (se puede usar la función hash para calcular su hash; de lo contrario, se generará un TypeError).  
  ·No existe un orden específico para las claves.  

```python
# Crear y rellenar con valores
stock = {'eggs': 5, 'milk': 2}

# O crear un diccionario vacío
dictionary = {}

# Y rellenarlo después
dictionary['eggs'] = 5
dictionary['milk'] = 2

# Los valores también pueden ser listas
mydict = {'a': [1, 2, 3], 'b': ['one', 'two', 'three']}

# Usar el método list.append() para añadir nuevos elementos a la lista de valores
mydict['a'].append(4) # => {'a': [1, 2, 3, 4], 'b': ['uno', 'dos', 'tres']}
mydict['b'].append('four') # => {'a': [1, 2, 3, 4], 'b': ['uno', 'dos', 'tres', 'cuatro']}

# También podemos crear un diccionario usando una lista de tuplas de dos elementos:
iterable = [('eggs', 5), ('milk', 2)]
dictionary = dict(iterables)

# O usando un argumento de palabra clave:
dictionary = dict(eggs=5, milk=2)

# Otra forma sería usar dict.fromkeys:
dictionary = dict.fromkeys((milk, eggs)) # => {'leche': None, 'huevos': None}
dictionary = dict.fromkeys((milk, eggs), (2, 5)) # => {'leche': 2, 'huevos': 5}
```

# Sección 19.9: Creación de un diccionario ordenado

Puedes crear un diccionario ordenado que seguirá un orden determinado al iterar sobre las claves del diccionario.

Utilice OrderedDict del módulo collections. Al iterar sobre él, siempre devolverá los elementos del diccionario en el orden de inserción original.

```python
from collections import OrderedDict

d = OrderedDict()
d['first'] = 1
d['second'] = 2
d['third'] = 3
d['last'] = 4

# Outputs "first 1", "second 2", "third 3", "last 4"
for key in d:
    print(key, d[key])
```

# Sección 19.10: Desempaquetado de diccionarios mediante el operador **

Puedes usar el operador de desempaquetado de argumentos de palabra clave ** para pasar los pares clave-valor de un diccionario a los argumentos de una función. Un ejemplo simplificado de la [documentación oficial](https://docs.python.org/3/tutorial/controlflow.html#unpacking-argument-lists):

```python
>>>
>>> def parrot(voltage, state, action):
... print("This parrot wouldn't", action, end=' ')
... print("if you put", voltage, "volts through it.", end=' ')
... print("E's", state, "!")
...
>>> d = {"voltage": "four million", "state": "bleedin' demised", "action": "VOOM"}
>>> parrot(**d)

This parrot wouldn't VOOM if you put four million volts through it. E's bleedin' demised !
```

A partir de Python 3.5, también puedes usar esta sintaxis para fusionar un número arbitrario de objetos de tipo `dict`.

```python
>>> fish = {'name': "Nemo", 'hands': "fins", 'special': "gills"}
>>> dog = {'name': "Clifford", 'hands': "paws", 'color': "red"}
>>> fishdog = {**fish, **dog}
>>> fishdog

{'hands': 'paws', 'color': 'red', 'name': 'Clifford', 'special': 'gills'}
```

Como demuestra este ejemplo, las claves duplicadas se asignan a su último valor (por ejemplo, "Clifford" sobrescribe a "Nemo").

# Sección 19.12: El constructor dict()

El constructor `dict()` se puede usar para crear diccionarios a partir de argumentos con nombre, o a partir de un único iterable de pares clave-valor, o a partir de un único diccionario y argumentos con nombre.

```python
dict(a=1, b=2, c=3)                     # {'a': 1, 'b': 2, 'c': 3}
dict([('d', 4), ('e', 5), ('f', 6)])    # {'d': 4, 'e': 5, 'f': 6}
dict([('a', 1)], b=2, c=3)              # {'a': 1, 'b': 2, 'c': 3}
dict({'a' : 1, 'b' : 2}, c=3)           # {'a': 1, 'b': 2, 'c': 3}
```

# Sección 19.14: Todas las combinaciones de valores de diccionario

```python
options = {
    "x": ["a", "b"],
    "y": [10, 20, 30]
}
```

Dado un diccionario como el que se muestra arriba, donde hay una lista que representa un conjunto de valores para explorar para la
clave correspondiente. Supongamos que queremos explorar "x"="a" con "y"=10, luego "x"="a" con "y"=10, y así sucesivamente hasta que
hayamos explorado todas las combinaciones posibles.

Puedes crear una lista que devuelva todas esas combinaciones de valores utilizando el siguiente código.

```python
import itertools

options = {
    "x": ["a", "b"],
    "y": [10, 20, 30]}

keys = options.keys()
values = (options[key] for key in keys)
combinations = [dict(zip(keys, combination)) for combination in itertools.product(*values)]
print combinations
```

Esto nos da la siguiente lista almacenada en la variable combinations:

```python
[{'x': 'a', 'y': 10},
 {'x': 'b', 'y': 10},
 {'x': 'a', 'y': 20},
 {'x': 'b', 'y': 20},
 {'x': 'a', 'y': 30},
 {'x': 'b', 'y': 30}] 
```
