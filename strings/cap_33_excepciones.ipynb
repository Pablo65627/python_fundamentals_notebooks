## Sintaxis de la sentencia `try`

La sentencia `try` debe estar acompañada de `except` o de `finally`, y el orden de las partes debe ser el siguiente.

Recuerda que en la documentación sobre líneas de comandos y código en general los símbolos `[]` indican que ese argumento es opcional y no se requiere para que la sentencia se ejecute.

Además, el metacarácter `*` indica que la sentencia `except` puede aparecer múltiples veces:

```python
try:
    statements
except [type [as value]]:
    statements
[except [type [as value]]:
    statements]*
[else:
    statements]
[finally:
    statements]
```
o ninguna:

```python
try:
    statements
finally:
    statements
```

````python
try:
    statements                  # Ejecuta esta acción principal primero
except name1:
    statements                  # Se ejecuta si se genera name1 durante el bloque try  
except (name2, name3):
    statements                  # Ejecutar si ocurre alguna de estas excepciones
except name4 as var:
    statements                  # Ejecutar si se genera name4, asignar la instancia generada a var
except:
    statements                 # Ejecutar para todas las demás excepciones que se produzcan
    statements
    statements
else:
    statements                 # Ejecutar si no se produjo ninguna excepción durante el bloque try
````

## Ejemplo try

````python
sep = '-' * 45 + '\n'                         # Crea una línea de 45 guiones y añade un salto de línea   
print(sep + 'EXCEPTION RAISED AND CAUGHT')    # Imprime el título del ejemplo
try:
    x = 'spam'[99]                            # Intenta acceder a una posición que no existe en la cadena
except IndexError:
    print('except run')                       # Se ejecuta si se produce un error IndexError
finally:
    print('finally run')                      # Se ejecuta siempre, independientemente de si se produce una excepción
print('after run')                            # Se ejecuta después del bloque try-except-finally
````

````python
print(sep + 'NO EXCEPTION RAISED')            # Imprime el título del ejemplo   
try:
    x = 'spam'[3]                             # Accede a la posición 3 de la cadena ('m'), sin producir errores
except IndexError:
    print('except run')                       # Se ejecuta si se produce un error IndexError
finally:
    print('finally run')                      # Se ejecuta siempre, aunque no haya errores
print('after run')                            # Se ejecuta después de finalizar la estructura
````

````python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE') # Imprime el título del ejemplo
try:
    x = 'spam'[3]                             # Accede a la posición 3 de la cadena ('m'), sin producir errores        
except IndexError:
    print('except run')                       # Se ejecuta si se produce un error IndexError
else:
    print('else run')                         # Se ejecuta únicamente si no se produce ninguna excepción en try 
finally:
    print('finally run')                      # Se ejecuta siempre, haya ocurrido una excepción o no
print('after run')                            # Se ejecuta después de finalizar toda la estructura
````

````python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT') # Imprime el título del ejemplo
try:
    x = 1 / 0                                  # Intenta dividir entre cero, lo que provoca un error ZeroDivisionError
except IndexError:
    print('except run')                        # Solo captura errores IndexError, no captura ZeroDivisionError
finally:
    print('finally run')                       # Se ejecuta siempre, incluso si el error no ha sido capturado
print('after run')                             # No se ejecuta porque el error no ha sido capturado y el programa se interrumpe
````
