# Cap 19 diccionarios

Los items se almacenan y recuperar por una clave, en vez de por su posición en la estructura.

Es una tipo de agregación en el cual el nombre de los items tiene más significado que su posición.

Los diccionarios pueden reemplazar muchos de los algoritmos de búsqueda y estructuras de datos que deberías implementar manualmente en lenguajes de bajo nivel.



Características:

  ·Se accede a los elementos por su clave o key, no por la posición que ocupan.  
  ·Son colecciones no ordenadas de objetos de cualquier tipo, y el tipo de los elementos que lo forman puede ser distinto.  
  ·Poseen longitud variable y pueden anidarse.  
  ·Son mapas mutables.  
  ·Son tablas de referencias a objetos o hash tables.  

```python
# D = {key: value}
D = {'spam': 2, 'ham': 1, 'eggs': 3}        # construir un diccionario
D
```

Technically, the ordering is pseudo-random—it’s not truly random

```python
D['spam']           # recuperar un valor su clave
```

```python
len(D)      # Numero de entradas del diccionario
# 3
```

```python
'ham' in D      # Key membership test alternative
# True
```

```python


list(D.keys())      # Crear una lista con las claves del diccionario
# ['eggs', 'spam', 'ham']

```

Es necesario encerrar el objeto que devuelve `keys()` en el constructor `list` en Python 3.X porque `keys()` en 3.X devuelve un objeto iterable en vez de una lista física.

## Cambiando diccionarios in place

```python
D['ham'] = ['grill', 'bake', 'fry']         # Change entry (value=list)
D
# {'eggs': 3, 'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

```python
del D['eggs']               # Delete entry
D
# {'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

Al contrario que las listas, cuando asignas una nueva clave en un diccionario (una que no haya sido asignada con anterioridad) creas una nueva entrada en el diccioanrio.

```python
D['brunch'] = 'Bacon'       # Add new entry
D
# {'brunch': 'Bacon', 'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

#### Métodos de un objeto diccionario

`values`

El método `values` devuelve todos los valores del diccionario.

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.values())                
# [3, 2, 1]
```

`items`

`items` devuelve todas las tuplas `(key,value)`.

```python
list(D.items())
# [('eggs', 3), ('spam', 2), ('ham', 1)]
```

`get`

Fetching una clave que no existe es normalmente un error, pero el método `get` devuelve `None` como valor por defecto o un valor pasado como por defecto si la clave no existe.

```python
D['toast']      # KeyError: 'toast'   
```

```python
D.get('spam')
# 2
```

```python
print(D.get('toast'))       # Una clave que no existe
# None
```

```python
D.get('toast', 88)          # proporcionando un valor por defecto
# 88
```

`update`

El metodo `update` une las claves y valores de un diccionaroi en otro, sobreescribiendo valores de la misma clave si hay conflicto.

```python
# D = {'eggs': 3, 'spam': 2, 'ham': 1}
D2 = {'toast':4, 'muffin':5, 'ham': 10}
D.update(D2)
D
# {'eggs': 3, 'muffin': 5, 'toast': 4, 'spam': 2, 'ham': 10}
```

`pop`

`pop` elimina una clave de un diccionario y devuelve el valor asociado a la clave.

```python
D.pop('muffin')
# 5
D.pop('toast')
# 4
D
# {'eggs': 3, 'spam': 2, 'ham': 10}
```

#### Vistas diccionario y sets

La vista objeto devuelta por el método `keys` es como un `set` y soporta las operaciones con conjuntos, como la intersección y la unión.

```python
K, V
```

```python
K | {'x': 4}            # La vista `keys` es como un `set` 
```

La vista `values` no es como un `set`.

```python
V & {'x': 4}            # TypeError: unsupported operand type(s) for &: 'dict_values' and 'dict'
```

La vista `items` genera pares (key, value) que son únicos y hashable (inmutables).

```python
V & {'x': 4}.values()   # TypeError: unsupported operand type(s) for &: 'dict_values' and 'dict_values'
```

### Sección 19.1: Introducción a los diccionarios

Un diccionario es un ejemplo de estructura de almacenamiento de pares clave-valor, también conocida como Mapping en Python. Permite guardar y recuperar elementos haciendo referencia a una clave. Como se accede a los diccionarios mediante claves, las búsquedas son muy rápidas. En las versiones de Python descritas en este capítulo, no se consideran estructuras ordenadas. Crear un diccionario. Los diccionarios se pueden inicializar de varias maneras.

### Sintaxis literal

```python
d = {} # empty dict
d = {'key': 'value'} # dict with initial values
```

Python 3.x

```python
# Also unpacking one or multiple dictionaries with the literal syntax is possible
# makes a shallow copy of otherdict
d = {**otherdict}
# also updates the shallow copy with the contents of the yetanotherdict.
d = {**otherdict, **yetanotherdict}
```

### Comprensión de diccionarios

```python
d = {k:v for k,v in [('key', 'value',)]}
```

Véase también: comprensiones.

### Clase integrada dict()

```python
d = dict() # empty dict
d = dict(key='value') # explicit keyword arguments
d = dict([('key', 'value')]) # passing in a list of key/value pairs
# make a shallow copy of another dict (only possible if keys are only strings!)
d = dict(**otherdict)
```

### Modificar un diccionario

Para añadir elementos, crea una clave nueva y asígnale un valor:

```python
d['newkey'] = 42
```

También se pueden utilizar listas y diccionarios como valores:

```python
d['new_list'] = [1, 2, 3]
d['new_dict'] = {'nested_dict': 1}
```

Para eliminar un elemento, elimina su clave:

```python
del d['newkey']
```

### Sección 19.2: Evitar excepciones KeyError

Un error habitual es intentar acceder a una clave que no existe. Esto suele provocar una excepción KeyError.

```python
mydict = {}
mydict['not there']
Traceback (most recent call last):
File "<stdin>", line 1, in <module>
KeyError: 'not there'
```

Una forma de evitarlo es utilizar dict.get, que permite indicar un valor predeterminado para devolver si la clave no existe.

```python
value = mydict.get(key, default_value)
```

Devuelve mydict[key] si existe; si no, devuelve default_value. Esto no añade la clave al diccionario. Si quieres guardar ese par clave-valor, usa mydict.setdefault(key, default_value).

```python
[ ]:
mydict = {}
print(mydict)
# {}
print(mydict.get("foo", "bar"))
# bar
print(mydict)
# {}
print(mydict.setdefault("foo", "bar"))
# bar
print(mydict)
# {'foo': 'bar'}
```

Otra alternativa es capturar la excepción:

```python
try:
value = mydict[key]
except KeyError:
value = default_value
```

También puedes comprobar si la clave está en el diccionario:

```python
if key in mydict:
value = mydict[key]
else:
value = default_value
```

En entornos con varios hilos, la clave podría eliminarse después de comprobar que existe. Esto crea una condición de carrera y la excepción aún podría producirse. Otra opción es collections.defaultdict, una subclase de dict que utiliza default_factory para crear entradas al acceder a claves nuevas.

### Sección 19.3: Recorrer un diccionario.

Si utilizas un diccionario como iterador (por ejemplo, en un bucle for), se recorren sus claves.

```python
d = {'a': 1, 'b': 2, 'c':3}
for key in d:
print(key, d[key])
# c 3
# b 2
# a 1
```
